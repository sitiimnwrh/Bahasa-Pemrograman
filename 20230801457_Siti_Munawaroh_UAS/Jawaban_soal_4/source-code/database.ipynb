```python
import pymysql

# Koneksi ke database MySQL
connection = pymysql.connect(
    host='localhost',
    user='root',
    password='password',
    database='test_db'
)

try:
    with connection.cursor() as cursor:
        # Membuat tabel
        cursor.execute('''
        CREATE TABLE IF NOT EXISTS Users (
            UserID INT AUTO_INCREMENT PRIMARY KEY,
            Name VARCHAR(100),
            Email VARCHAR(100) UNIQUE,
            DateOfBirth DATE
        )
        ''')
        connection.commit()

        # Menambahkan data
        cursor.execute('''
        INSERT INTO Users (Name, Email, DateOfBirth)
        VALUES (%s, %s, %s)
        ''', ('Alice Smith', 'alice@example.com', '1990-05-12'))
        connection.commit()

        # Mengambil data
        cursor.execute('SELECT * FROM Users')
        rows = cursor.fetchall()
        for row in rows:
            print(row)
finally:
    connection.close()
```
