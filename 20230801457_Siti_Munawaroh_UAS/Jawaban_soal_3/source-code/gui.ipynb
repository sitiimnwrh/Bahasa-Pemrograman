```python
import tkinter as tk

def greet():
    print("Hello, world!")

# Fungsi untuk menjalankan Tkinter di latar belakang
def run_gui():
    window = tk.Tk()
    window.title("Contoh GUI Sederhana")

    label = tk.Label(window, text="Selamat Datang di GUI dengan Tkinter!")
    label.pack(pady=10)

    button = tk.Button(window, text="Klik Saya", command=greet)
    button.pack(pady=10)

    window.mainloop()

# Menjalankan Tkinter di thread terpisah
import threading
gui_thread = threading.Thread(target=run_gui)
gui_thread.start()
```
